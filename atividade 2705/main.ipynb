### 1. Escreva uma função recursiva que receba uma árvore binária ab e um número x e remova da árvore todas as folhas que tenham item igual a x. 

```python
def remove_folha_x(root, x):
    # se a árvore estiver vazia
    if root is None:
        return None

    if root.left is None and root.right is None and root.value == x:
        # se a folha é igual a x
        return None if root.value == x else root

    # se a folha não é igual a x
    root.left = remove_folha_x(root.left, x)
    root.right = remove_folha_x(root.right, x)

    return root
```

### 2. Escreva uma função não recursiva que calcule o número de nós de uma árvore binária.

```python
def conta_folhas(root):
    # se a árvore estiver vazia
    if root is None:
        return 0

    # pilha para armazenar as folhas e o contador
    stack = [root]
    count = 0

    # enquanto não vazia
    while stack:
        node = stack.pop() # pega o último nó da pilha

        # se o nó é uma folha e não tem filhos
        if node.left is None and node.right is None:
            count += 1

        else:
            if node.right:
                stack.append(node.right)
            if node.left:
                stack.append(node.left)

    return count
```

### 3. Escreva uma função que calcule o número de folhas de uma árvore binária. Faça três versões: uma que percorra a árvore em in-ordem, outra que percorra a árvore em pre-ordem e outra que percorra a árvore em post-ordem.

```python
def inorder(root):
    def inorder(node):
        # se a árvore estiver vazia
        if node is None:
            return 0

        # soma folhas da subárvore esquerda
        total = inorder(node.left)

        # conta esse nó se for folha
        if node.left is None and node.right is None:
            total += 1

        # soma folhas da subárvore direita
        total += inorder(node.right)
        return total

    return inorder(root)


def preorder(root):
    # se a árvore estiver vazia
    if root is None:
        return 0

    # conta esse nó se for folha
    count = 1 if (root.left is None and root.right is None) else 0

    # soma folhas da subárvore esquerda
    count += preorder(root.left)

    # soma folhas da subárvore direita
    count += preorder(root.right)
    return count


def postorder(root):
    # se a árvore estiver vazia
    if root is None:
        return 0

    # soma folhas das subárvores
    count = postorder(root.left) + postorder(root.right)

    # conta esse nó se for folha
    if root.left is None and root.right is None:
        count += 1

    return count
```

### 4. Pesquisar como são realizadas as operações Inserir e Remover nas Árvores Rubro-Negras.

#### Inserção
        
Insere o novo nó como em uma BST normal, pintado de vermelho.

Se o pai também for vermelho, viola a propriedade “pai vermelho” e aplica rotações (simples ou dupla) e/ou recoloração entre tio, pai e avô até restaurar propriedades da rubro-negra (raiz preta, nenhum caminho com dois vermelhos consecutivos, uniformidade de nós pretos em todo caminho).

#### Remoção

Remoção inicial como em BST (se for interno, troca com anterior/sucessor).

Se o nó removido (ou substituto) for preto, pode perder a contagem de pretos em algum caminho, o que é o chamado “nó duplamente preto".

Aplica-se uma sequência de rotações e/ou recolorações, considerando o “irmão” do nó em falta, até redistribuir os pretos corretamente e restaurar as propriedades.
